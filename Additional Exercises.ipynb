```python
# Student Parameters
color_line = '#1598C1'

try:
    # Calculations
    yearly_counts = df.groupby('year').size().reset_index(name='count')
    yearly_counts['cumulative'] = yearly_counts['count'].cumsum()

    # Visualization
    fig, ax1 = plt.subplots(figsize=(12, 6))

    # Primary Axis: Yearly Bar Chart
    sns.barplot(x='year', y='count', data=yearly_counts, color=color_bar, alpha=0.6, ax=ax1)
    ax1.set_ylabel("Songs Released Per Year", color=color_bar)

    # Secondary Axis: Cumulative Line Chart
    ax2 = ax1.twinx()
    sns.lineplot(x=range(len(yearly_counts)), y='cumulative', data=yearly_counts, color=color_line, marker='o', ax=ax2)
    ax2.set_ylabel("Cumulative Total", color=color_line)

    plt.title(f"{student_name} ({student_id})\nSong Release Trends Over Time")
    ax1.set_xticklabels(ax1.get_xticklabels(), rotation=45)
    plt.show()

except Exception as e:
    print(f"Error: {e}")
```
